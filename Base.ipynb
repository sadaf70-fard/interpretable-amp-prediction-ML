```python
import os
import numpy as np
import pandas as pd
from collections import Counter
from Bio.SeqUtils.ProtParam import ProteinAnalysis

# Amino acid list
AMINO_ACIDS = 'ACDEFGHIKLMNPQRSTVWY'

# Read sequences from a txt file
def read_sequences(file_path):
    sequences = []
    with open(file_path, 'r') as file:
        lines = file.readlines()

    current_id = None
    current_seq = []

    for line in lines:
        line = line.strip()
        if not line:
            continue  # Skip empty lines
        if line.startswith('>'):
            if current_id and current_seq:
                sequences.append((current_id, ''.join(current_seq)))
            current_id = line[1:].strip()
            current_seq = []
        else:
            current_seq.append(line.strip())

    if current_id and current_seq:
        sequences.append((current_id, ''.join(current_seq)))

    return sequences


# Amino Acid Composition (AAC)
def compute_aac(sequence):
    L = len(sequence)
    counts = Counter(sequence)
    return [counts[aa] / L if L > 0 else 0 for aa in AMINO_ACIDS]

# Dipeptide Composition (DPC)
def compute_dpc(sequence):
    if len(sequence) < 2:
        return [0] * 400
    dipeptides = [a + b for a in AMINO_ACIDS for b in AMINO_ACIDS]
    total = len(sequence) - 1
    counts = Counter([sequence[i:i + 2] for i in range(total)])
    return [counts[dp] / total if total > 0 else 0 for dp in dipeptides]

#  Tripeptide Composition (TPC)
def compute_tpc(sequence):
    if len(sequence) < 3:
        return [0] * 8000
    tripeptides = [a + b + c for a in AMINO_ACIDS for b in AMINO_ACIDS for c in AMINO_ACIDS]
    total = len(sequence) - 2
    counts = Counter([sequence[i:i + 3] for i in range(total)])
    return [counts[tp] / total if total > 0 else 0 for tp in tripeptides]

#  Binary Encoding (One-hot)
def one_hot_encode(seq, max_len=30):
    aa_to_index = {a: i for i, a in enumerate(AMINO_ACIDS)}
    mat = np.zeros((max_len, 20))
    for i, c in enumerate(seq[:max_len]):
        if c in aa_to_index:
            mat[i, aa_to_index[c]] = 1
    return mat.flatten()

#  Physicochemical Properties
def compute_pcp(sequence, table=property_table):
    features = []

    for prop_name, aa_values in table.items():
        values = []
        for aa in sequence:
            if aa in aa_values:
                values.append(aa_values[aa])
        avg = sum(values) / len(values) if values else 0
        features.append(avg)

    return features

# Extract features from all .txt files in a folder
def extract_features_from_folder(folder_path, label=1, max_len=30):
    all_features = []
    all_labels = []
    all_seqs = []                                   # +

    for file_name in os.listdir(folder_path):
        if file_name.endswith('.txt'):
            file_path = os.path.join(folder_path, file_name)
            print(f" Processing file: {file_path}")
            sequences = read_sequences(file_path)

            for peptide_id, sequence in sequences:
                try:
                    sequence = sequence.upper()
                    if not sequence.isalpha() or any(aa not in AMINO_ACIDS for aa in sequence):
                        print(f"Invalid or non-standard sequence: {peptide_id}")
                        continue

                    aac = compute_aac(sequence)
                    dpc = compute_dpc(sequence)
                    tpc = compute_tpc(sequence)
                    pcp = compute_pcp(sequence)
                    be = one_hot_encode(sequence, max_len)

                    features = aac + dpc + tpc + pcp + list(be)
                    all_features.append(features)
                    all_labels.append(label)
                    all_seqs.append(sequence)       # +

                except Exception as e:
                    print(f" Error processing sequence {peptide_id}: {e}")

    return np.array(all_features), np.array(all_labels), all_seqs

# Main function to process both AMP and non-AMP
def main():
    amp_folder = os.path.join("data", "AMP")
    non_amp_folder = os.path.join("data", "Non_AMP")

    print("Extracting AMP features...")
    amp_features, amp_labels, amp_seqs = extract_features_from_folder(amp_folder, label=1)

    print("Extracting non-AMP features...")
    non_amp_features, non_amp_labels, non_amp_seqs = extract_features_from_folder(non_amp_folder, label=0)

    # Combine all
    all_features = np.vstack([amp_features, non_amp_features])
    all_labels = np.concatenate([amp_labels, non_amp_labels])
    all_seqs = amp_seqs + non_amp_seqs              # +

    df = pd.DataFrame(all_features)
    df['sequence'] = all_seqs
    df['label'] = all_labels
    df.to_csv('amp_nonamp_combined.csv', index=False)

    print("All features extracted and saved to 'amp_nonamp_combined.csv'")

if __name__ == '__main__':
    main()
```

<h3> Random Undersampling of the Majority Class

```python
import pandas as pd

df = pd.read_csv('amp_nonamp_combined.csv')
print("before dedup:", df.shape)

df = df.drop_duplicates(subset='sequence').reset_index(drop=True)
print("after dedup :", df.shape, " unique sequences:", df['sequence'].nunique())
print(df['label'].value_counts())

class_counts = df['label'].value_counts()
min_class = class_counts.idxmin()
min_count = class_counts.min()

df_balanced = df.groupby('label').sample(n=min_count, random_state=42)

df_balanced = df_balanced.sample(frac=1, random_state=42).reset_index(drop=True)

df_balanced.to_csv('balanced_dataset.csv', index=False)

print("Balanced dataset saved with shape:", df_balanced.shape)
```

```python
from Bio import Align

aligner = Align.PairwiseAligner(mode="local", match_score=1, mismatch_score=-1,
                                open_gap_score=-1, extend_gap_score=-0.5)

def identity(a, b):
    short = min(len(a), len(b))
    if aligner.score(a, b) < 0.4 * short:       # cheap check, skips most pairs
        return 0.0
    aln = aligner.align(a, b)[0]
    matches = sum(1 for (s1, e1), (s2, e2) in zip(*aln.aligned)
                    for k in range(e1 - s1) if a[s1 + k] == b[s2 + k])
    return matches / short

seqs = df_balanced["sequence"].tolist()
order = sorted(range(len(seqs)), key=lambda i: -len(seqs[i]))   # longest first

reps, cluster_of = [], {}
for n, i in enumerate(order):
    s = seqs[i]
    for c, r in enumerate(reps):
        if identity(s, r) >= 0.4:
            cluster_of[i] = c
            break
    else:
        reps.append(s)
        cluster_of[i] = len(reps) - 1
    if (n + 1) % 200 == 0:
        print(n + 1, len(reps), flush=True)

df_balanced["cluster"] = [cluster_of[i] for i in range(len(seqs))]
df_balanced.to_csv("balanced_dataset.csv", index=False)

sizes = df_balanced["cluster"].value_counts()
print(len(seqs), "sequences ->", len(sizes), "clusters")
print(sizes.describe())
print("ten largest:", sizes.head(10).tolist())
```

```python
from Bio import Align
import numpy as np

aligner = Align.PairwiseAligner(mode="local", match_score=1, mismatch_score=-1,
                                open_gap_score=-1, extend_gap_score=-0.5)

def identity(a, b):
    short = min(len(a), len(b))
    if aligner.score(a, b) < 0.4 * short:
        return 0.0
    aln = aligner.align(a, b)[0]
    matches = sum(1 for (s1, e1), (s2, e2) in zip(*aln.aligned)
                    for k in range(e1 - s1) if a[s1 + k] == b[s2 + k])
    return matches / short

# the split being diagnosed: random, or cluster-aware via GroupShuffleSplit
from sklearn.model_selection import train_test_split
train_idx, test_idx = train_test_split(
    np.arange(len(df)), test_size=0.2, stratify=df["label"], random_state=42)

train_seqs = df["sequence"].iloc[train_idx].tolist()
test_seqs  = df["sequence"].iloc[test_idx].tolist()

max_identity = []
for n, s in enumerate(test_seqs):
    best = 0.0
    for r in train_seqs:
        best = max(best, identity(s, r))
        if best >= 1.0:
            break
    max_identity.append(best)
    if (n + 1) % 100 == 0:
        print(n + 1, f"{sum(m >= 0.4 for m in max_identity)} above 0.4", flush=True)

max_identity = np.array(max_identity)
print(f"\ntest sequences with a training match >= 0.4: "
      f"{(max_identity >= 0.4).sum()} / {len(max_identity)} "
      f"({100*(max_identity >= 0.4).mean():.1f}%)")
print("max identity percentiles:", np.percentile(max_identity, [50, 75, 90, 95, 100]).round(3))
```

```python
import re, urllib.request
import pandas as pd

URL = ("https://raw.githubusercontent.com/BioGenies/AMPBenchmark/"
       "main/data/AMPBenchmark_public.fasta")
urllib.request.urlretrieve(URL, "AMPBenchmark_public.fasta")

# --- read ---
rows, head = [], None
for line in open("AMPBenchmark_public.fasta"):
    line = line.strip()
    if not line:
        continue
    if line.startswith(">"):
        head = line[1:]
    else:
        m = re.search(r"sampling_method=(.+?)_AMP=", head)
        rows.append({"sequence": line.upper(),
                     "label": 1 if "AMP=1" in head else 0,
                     "method": m.group(1) if m else "positive"})

bench = pd.DataFrame(rows)
bench = bench[bench["sequence"].str.fullmatch("[ACDEFGHIKLMNPQRSTVWY]{3,}")]
bench = bench.drop_duplicates(subset="sequence")
print("unique sequences in AMPBenchmark:", len(bench))

# --- drop anything our model has already seen ---
ours = set(df["sequence"])
before = len(bench)
bench = bench[~bench["sequence"].isin(ours)]
print(f"removed {before - len(bench)} sequences overlapping our dataset")

# --- balance ---
NEG_METHOD = "Gabere&Noble"        # one published negative-sampling method

pos = bench[bench["label"] == 1]
neg = bench[(bench["label"] == 0) & (bench["method"] == NEG_METHOD)]
print(f"available: {len(pos)} positives, {len(neg)} negatives from {NEG_METHOD}")

n = min(len(pos), len(neg))
external = pd.concat([pos.sample(n, random_state=42),
                      neg.sample(n, random_state=42)])
external = external.sample(frac=1, random_state=42).reset_index(drop=True)

external.to_csv("external_benchmark.csv", index=False)
print("external_benchmark.csv:", external.shape)
print(external["label"].value_counts())
```
